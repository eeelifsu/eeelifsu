```python
# 1. Öğrenciler ve sınav notları analizi
# 2. Veri oluşturalım - 2 değişken olsun - 10 öğrenciler(isimleri elle ver) - 3 adet notu olsun  notlar(otomatik olustur)
# 3. Veri oluşturalımNot ortalamalarını yazdır satır satır her öğrenci içinmiş gibi (10 a 3 matris olacak mantıken)
# 4. her sınavın ortalamasını
# 5. sınıf geneli istatistik (en yüksek en düşük genel ort gibi)
# 6. where ile harf notu ve geçti geçemedi gibi
```

```python
import numpy as np

ogrenciler = np.array([
    "Elifsu",
    "Sinan",
    "Berkay",
    "Muhammet",
    "Melike",
    "Onur",
    "Hüseyin"
])

notlar = np.random.randint(0, 101, (7, 3))
notlar

ortalamalar = np.mean(notlar, axis=1)
ortalamalar

for i in range(len(ogrenciler)):
    print(ogrenciler[i], ":", round(ortalamalar[i], 2))


sinav_ortalamalari = np.mean(notlar, axis=0)

print("1. Sınav ortalaması:", sinav_ortalamalari[0])
print("2. Sınav ortalaması:", sinav_ortalamalari[1])
print("3. Sınav ortalaması:", sinav_ortalamalari[2])


en_yuksek = np.max(notlar)
en_dusuk = np.min(notlar)
sinif_ortalamasi = np.mean(notlar)

print("En yüksek not:", en_yuksek)
print("En düşük not:", en_dusuk)
print("Sınıf genel ortalaması:", sinif_ortalamasi)

sonuc = np.where(ortalamalar >= 50, "Geçti", "Kaldı")

for i in range(len(ogrenciler)):
    print(ogrenciler[i], ":", sonuc[i])
```
